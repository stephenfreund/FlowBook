# Progress: Unstable Write Sets (potential non-termination)

**Property:** a cell that writes a *different set of variables*
on different runs can keep marking earlier cells stale forever —
rerunning stale cells would never certify this notebook.  Instead
of looping, FlowBook reports **potential non-termination** at the
unstable cell.

Cell B writes `low` on some runs and `high` on others (cell A's
`phase` deliberately flips on every execution, standing in for
`random.random() > 0.5`).  When B's rerun drops a write, the
variable's last writer (cell A) is marked stale; A's rerun marks
B stale again; and B flips again.

**Try it:**
1. Run all cells — the first pass comes up clean.
2. Edit cell A (change `high = 100` to `high = 101`), then run
   all stale cells: the sweep runs A, B, A, B and stops with a
   potential non-termination report at cell B, instead of
   cycling forever.


```python
# Cell A: phase flips on every execution (untracked module state,
# a deterministic stand-in for a random coin flip)
import builtins
phase = getattr(builtins, "_progress_demo_phase", False)
builtins._progress_demo_phase = not phase
low = 0
high = 100
```

```python
# Cell B: writes a DIFFERENT variable depending on phase —
# this is the unstable cell the report will name
if phase:
    low = 101
else:
    high = 90
```

```python
# Cell C: reads low and high
span = high - low
print(f"span = {span}")
```
