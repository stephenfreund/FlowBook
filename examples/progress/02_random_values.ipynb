# Progress: Random Values Are Fine

**Property:** nondeterministic *values* never block progress.
Cell A draws a fresh random sample on every run, so downstream
results differ between reruns — but it always writes the *same
set of variables*, and FlowBook's rerun check compares only the
read/write sets, never values.

**Try it:**
1. Run all cells.
2. Edit cell A (e.g. change the sample size), then run all stale
   cells.  Every rerun draws new values; the sweep terminates
   clean with no warning, every time.


```python
# Cell A: a fresh random sample on every run — same variables written
import random
sample = [random.gauss(0, 1) for _ in range(1000)]
```

```python
# Cell B: reads sample, writes sample_mean
sample_mean = sum(sample) / len(sample)
```

```python
# Cell C: reads sample and sample_mean, writes sample_var
sample_var = sum((v - sample_mean) ** 2 for v in sample) / len(sample)
```

```python
# Cell D: reads the statistics
print(f"mean={sample_mean:.3f}  var={sample_var:.3f}")
```
