# Progress: Recreating a Random DataFrame Is Fine

**Property:** rerunning `df = pd.DataFrame(...)` allocates a *new
object* with new random contents on every run.  Object identity
and values both change, but the columns read and written stay the
same, so the rerun check stays quiet and the sweep terminates.

**Try it:**
1. Run all cells.
2. Edit cell A (e.g. change the number of rows), then run all
   stale cells — clean again, no warning.


```python
# Cell A: a brand-new DataFrame with random contents on every run
import random
import pandas as pd
df = pd.DataFrame({
    "price": [random.uniform(1, 10) for _ in range(100)],
    "qty": [random.randint(1, 5) for _ in range(100)],
})
```

```python
# Cell B: reads df's price and qty columns, writes revenue
revenue = (df["price"] * df["qty"]).sum()
```

```python
# Cell C: reads revenue
print(f"total revenue = {revenue:.2f}")
```
