# Progress: Random Value Feeding a Branchy Write Set

**Property:** randomness in *values* is fine (notebook 02), and
branch-dependent *write sets* are fine when their inputs are
stable — but the combination is not: cell B writes `grade_a` or
`grade_b` depending on a random score, so its recorded results
can never be trusted to reproduce.  When a rerun actually flips
the branch, FlowBook warns at B and continues; the sweep stops
with a potential non-termination error only if a cell exceeds
10 runs in one sweep.

**Try it:**
1. Run all cells.
2. Edit cell A and run all stale cells.  Repeat a few times:
   whenever the score crosses 0.5 between B's reruns, the sweep
   warns at cell B and keeps going until it converges.  (With a fair coin this
   happens within a couple of tries; notebook 10 shows the same
   failure deterministically.)

**The fix** is to make B's write set stable, e.g. write both
variables on every run:
`grade_a, grade_b = (1, 0) if score > 0.5 else (0, 1)`.


```python
# Cell A: a random score — plus both grade defaults
import random
score = random.random()
grade_a = 0
grade_b = 0
```

```python
# Cell B: which variable gets written depends on the random score
if score > 0.5:
    grade_a = 1
else:
    grade_b = 1
```

```python
# Cell C: reads both grades
total = grade_a + grade_b
print(f"total = {total}")
```
