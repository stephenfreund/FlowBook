# Progress: Deterministic Pipeline (rerun always terminates)

**Property:** running stale cells top to bottom terminates for
deterministic notebooks — each cell reruns at most once per sweep,
and FlowBook never reports a warning.

**Try it:**
1. Run all cells — the notebook becomes clean.
2. Edit cell A (change the prices) — cells B, C, D become stale.
3. Run all stale cells (or `run_actionable_cells` from MCP): the
   staleness cascades forward, every cell reruns exactly once, and
   the notebook is clean again.


```python
# Cell A: input data.  Edit this list, then rerun stale cells.
prices = [1.0, 2.0, 3.0, 4.0]
```

```python
# Cell B: reads prices, writes total
total = sum(prices)
```

```python
# Cell C: reads prices and total, writes mean
mean = total / len(prices)
```

```python
# Cell D: reads mean
print(f"mean price = {mean:.2f}")
```
