# FlowBook: Getting Started

This notebook demonstrates the basics of how FlowBook keeps a notebook *output consistent*: its recorded outputs match what a top-to-bottom run of the current cells would produce.

**Key concepts:**
- FlowBook tracks which variables each cell reads and writes
- It detects when cells become "stale" (need re-execution)
- It blocks cell runs that violate *rerun consistency*, such as backward mutations

**Instructions:** Run the cells in order A -> B -> C, then try running them out of order.

## Cell A: Define a variable

```python
x = 10
```

## Cell B: Use the variable

```python
y = x * 2
print(f"y = {y}")
```

## Cell C: Use the result

```python
z = y + 5
print(f"z = {z}")
```

---

## Try it yourself!

1. **Run A -> B -> C in order**: All cells should be clean (green checkmark)

2. **Change x to 20 in Cell A and rerun just Cell A**: Cell B becomes stale (needs rerun)

3. **Add a new Cell D: `x = 30` and run it**: This is a rerun consistency violation.
